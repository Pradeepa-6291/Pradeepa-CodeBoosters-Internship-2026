**1. 5 Vs of Big Data with Flipkart Case Study**
**Volume**

Volume refers to the enormous amount of data generated every day. Flipkart handles millions of customer transactions, product listings, reviews, clicks, and searches. During major sales events such as Big Billion Days, the platform generates terabytes of data within hours. Traditional databases struggle to process such large datasets efficiently, making big data technologies essential for storage and analysis.

**Velocity**

Velocity represents the speed at which data is generated and processed. Flipkart receives continuous streams of data from website visits, mobile app interactions, order placements, and payment gateways. Real-time processing helps Flipkart update inventory levels, recommend products instantly, and detect fraudulent transactions without delays.

**Variety**

Variety refers to different forms of data. Flipkart manages structured data such as customer records and orders, semi-structured data such as JSON logs, and unstructured data such as product images, videos, and customer reviews. Big data systems help integrate and analyze these diverse data sources effectively.

**Veracity**

Veracity focuses on data quality and accuracy. Flipkart receives data from multiple sellers, customers, and third-party logistics providers. Some information may contain errors, duplicates, or inconsistencies. Data cleaning and validation processes ensure reliable analytics and better business decisions.

**Value**

Value refers to extracting meaningful insights from data. Flipkart analyzes customer behavior, purchase history, and product trends to improve recommendations, optimize pricing, and increase customer satisfaction. The ultimate goal of big data is to create business value and drive growth.

**Transformations vs Actions in PySpark :**
**Transformations**
1.Transformations are operations that
create a new DataFrame or RDD from an existing one.
2.They do not execute immediately and follow lazy execution.
3.Spark only builds an execution plan when transformations are applied.
4.Multiple transformations can be chained together before execution.
5.They improve performance by allowing Spark to optimize the execution plan.
6.The output of a transformation is another DataFrame or RDD.

Common examples:
select()
filter()
groupBy()
join()
withColumn()

**Actions**
1.Actions are operations that trigger the execution of transformations.
2.They perform the actual computation and return results.
3.Actions can return a value to the driver program or display data.
4.They execute the execution plan created by transformations.
5.The output can be a number, a collection of records, or displayed data.

Common examples:
show()
count()
collect()
first()
take()

**EXAMPLE:**

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Example").getOrCreate()

data = [("Aarav",50000),("Priya",60000)]

df = spark.createDataFrame(data,["name","salary"])

filtered_df = df.filter(df.salary > 55000)

filtered_df.show()
```

**Top 5 Cities by Revenue Using PySpark DataFrame API**

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum

spark = SparkSession.builder.appName("RevenueAnalysis").getOrCreate()

data = [
("Mumbai",120000),
("Delhi",95000),
("Bangalore",150000),
("Chennai",80000),
("Hyderabad",110000),
("Mumbai",130000),
("Bangalore",125000),
("Delhi",105000)
]

df = spark.createDataFrame(data,["city","revenue"])

top_cities = df.groupBy("city") \
               .agg(sum("revenue").alias("total_revenue")) \
               .orderBy("total_revenue", ascending=False) \
               .limit(5)

top_cities.show()
```

**Parquet vs CSV: Performance and Storage Comparison**

Parquet is widely used in big data environments because it stores data in a columnar format, reducing storage requirements and improving query performance. When analyzing only a few columns from a large dataset, Parquet reads only the required columns, whereas CSV must scan the entire file. This results in faster processing and lower resource consumption. CSV files are simple and human-readable but are less suitable for large-scale analytics.


**Medallion Architecture (Bronze, Silver, Gold)**

**Bronze Layer**

The Bronze layer contains raw data collected directly from source systems such as applications, databases, APIs, and log files. At this stage, data is stored in its original format without major transformations. The purpose of the Bronze layer is to preserve historical records and maintain a reliable source of truth for future processing.

**Silver Layer**

The Silver layer contains cleaned and validated data. Data engineers remove duplicates, handle missing values, standardize formats, and apply business rules. The Silver layer improves data quality and creates structured datasets that are ready for analysis and machine learning applications.

**Gold Layer**

The Gold layer contains highly refined and business-ready data. It includes aggregated metrics, KPIs, dashboards, and reporting datasets used by business analysts and decision-makers. For Flipkart, the Gold layer may contain daily sales summaries, top-selling products, customer segmentation reports, and revenue analytics.
