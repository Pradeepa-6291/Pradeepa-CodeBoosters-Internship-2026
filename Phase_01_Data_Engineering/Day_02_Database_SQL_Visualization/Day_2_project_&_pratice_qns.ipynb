```python
# STUDENT PERFORMANCE DASHBOARD USING CSV FILE
# Codeboosters Tech Internship

# IMPORT LIBRARIES
import pandas as pd
import matplotlib.pyplot as plt

# LOAD CSV FILE
# Upload student_performance.csv in Google Colab first

df = pd.read_csv("/content/student_performance.csv")

# DISPLAY FIRST 5 ROWS
print(df.head())

# PANEL 1 : AVG MATH SCORE BY DEPARTMENT

df1 = df.groupby('department')['math_score'].mean().round(1).reset_index()

# PANEL 2 : STUDENT COUNT BY DEPARTMENT

df2 = df.groupby('department').size().reset_index(name='total_students')

# PANEL 3 : TOP 8 STUDENTS BY TOTAL SCORE

df['total_score'] = (
    df['math_score'] +
    df['programming_score'] +
    df['programming_score'] +
    df['attendance_percentage']
)

df3 = df[['name', 'total_score']] \
        .sort_values(by='total_score', ascending=False) \
        .head(8)

# PANEL 4 : GENDER-WISE AVG ATTENDANCE

df4 = df.groupby('gender')['attendance_percentage'].mean().round(1).reset_index()

# CREATE DASHBOARD

fig, axes = plt.subplots(2,2, figsize=(16,11))

# DASHBOARD TITLE

fig.suptitle(
    "Student Performance Dashboard — Codeboosters Tech Internship",
    fontsize=20,
    fontweight='bold',
    color='midnightblue'
)

# PANEL 1 (TOP LEFT)
# AVG MATH SCORE BY DEPARTMENT

colors1 = ['skyblue','teal','orange','purple']

bars = axes[0,0].bar(
    df1['department'],
    df1['math_score'],
    color=colors1
)

axes[0,0].set_title(
    "Avg Math Score by Department",
    fontsize=13,
    fontweight='bold'
)

axes[0,0].set_ylabel("Average Score")

axes[0,0].grid(axis='y', linestyle='--', alpha=0.3)

# VALUE LABELS
for bar in bars:

    height = bar.get_height()

    axes[0,0].text(
        bar.get_x() + bar.get_width()/2,
        height + 1,
        f'{height}',
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

# PANEL 2 (TOP RIGHT)
# PIE CHART

pie_colors = ['#4faad5','#0d8b7d','#e67e22','#8e44ad']

axes[0,1].pie(
    df2['total_students'],
    labels=df2['department'],
    autopct='%1.0f%%',
    startangle=90,
    colors=pie_colors
)

axes[0,1].set_title(
    "Students per Department",
    fontsize=13,
    fontweight='bold'
)

# PANEL 3 (BOTTOM LEFT)
# HORIZONTAL BAR CHART

bars2 = axes[1,0].barh(
    df3['name'],
    df3['total_score'],
    color='skyblue'
)

axes[1,0].invert_yaxis()

axes[1,0].set_title(
    "Top 8 Students — Total Score",
    fontsize=13,
    fontweight='bold'
)

axes[1,0].set_xlabel("Total Score (out of 400)")

axes[1,0].grid(axis='x', linestyle='--', alpha=0.3)

# VALUE LABELS
for bar in bars2:

    width = bar.get_width()

    axes[1,0].text(
        width + 2,
        bar.get_y() + bar.get_height()/2,
        f'{int(width)}',
        va='center',
        fontsize=9,
        fontweight='bold'
    )

# PANEL 4 (BOTTOM RIGHT)
# AVG ATTENDANCE BY GENDER

bars3 = axes[1,1].bar(
    df4['gender'],
    df4['attendance_percentage'],
    color=['skyblue','orange'],
    width=0.5
)

axes[1,1].set_title(
    "Avg Attendance by Gender",
    fontsize=13,
    fontweight='bold'
)

axes[1,1].set_ylabel("Attendance %")

axes[1,1].set_ylim(0,105)

axes[1,1].grid(axis='y', linestyle='--', alpha=0.3)

# VALUE LABELS
for bar in bars3:

    height = bar.get_height()

    axes[1,1].text(
        bar.get_x() + bar.get_width()/2,
        height + 1,
        f'{height}%',
        ha='center',
        fontsize=11,
        fontweight='bold'
    )

# LAYOUT

plt.tight_layout(rect=[0,0,1,0.95])

# SHOW DASHBOARD

plt.show()

```

```python
dept_data={
    'dept_code':['CS','EC','ME',"CE"],
    'dept_name':['Computer Science','Electronics','Mechanical','Civil'],
    'hod':['Dr sharma','Dr Guru','Dr Hamlin','Dr Arjun'],
    'established':[1997,1996,1999,2000],
    'intake':[60,70,65,75]
}

df_dept=pd.DataFrame(dept_data)
df_dept
```

```python
import sqlite3

# Establish a connection to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
c = conn.cursor()

df_dept.to_sql('departments',
          conn,
          if_exists='replace',
          index=False   #do not write dataframe row number as a column
          )
print("created 'departments' table ")
print(df_dept.to_string(index=False))

dept_map={
    'Computer Science':'CS',
    'Electronics':'EC',
    'Mechanical':'ME',
    'Civil':'CE'
}

# Reload the original DataFrame to ensure 'department' column has full names
df = pd.read_csv("student_performance.csv")

# Create a new 'dept_code' column by mapping the 'department' names
df['dept_code'] = df['department'].map(dept_map)

# Save the DataFrame with the new 'dept_code' column to the 'students' table
df.to_sql('students',conn,if_exists='replace',index=False)
print("Updated 'students' table with 'dept_code' column.")

print("New schema for 'students' table after adding 'dept_code' column:")
c.execute("PRAGMA table_info(students)")
info = c.fetchall()
for col in info:
  print(col[1], col[2])
```

```python
def run_query(query):
    c.execute(query)
    results = c.fetchall()
    if results:
        # Print column headers
        headers = [description[0] for description in c.description]
        print(headers)
        # Print rows
        for row in results:
            print(row)
    else:
        print("No results to display.")
```

```python
def run_query(query):
    c.execute(query)
    results = c.fetchall()
    if results:
        headers = [description[0] for description in c.description]
        df_result = pd.DataFrame(results, columns=headers)
        display(df_result)
    else:
        print("No results to display.")

# LEFT JOIN
# All records from LEFT table + matching from RIGHT
query_left = """
SELECT
s.name,
s.department,
s.math_score,
d.dept_name,
d.hod

FROM students AS s

LEFT JOIN departments AS d
ON s.dept_code = d.dept_code
"""

run_query(query_left)



# DISPLAY ALL STUDENTS

q7 = "SELECT * FROM students"

run_query(q7)
```

```python
# INNER JOIN
# Only matching records from both tables

query_inner = """
SELECT
s.name,
s.math_score,
d.dept_name,
d.hod,
d.established

FROM students AS s

INNER JOIN departments AS d
ON s.dept_code = d.dept_code

ORDER BY s.math_score DESC
LIMIT 8
"""

run_query(query_inner)
```

```python
# RIGHT JOIN (Simulated for SQLite)

query_right_simulated = """
SELECT
s.name,
s.department,
s.math_score,
d.dept_name,
d.hod

FROM departments AS d

LEFT JOIN students AS s
ON s.dept_code = d.dept_code
"""

run_query(query_right_simulated)

```

```python
# FULL OUTER JOIN (Simulated for SQLite)

query_full_outer_simulated = """
SELECT
s.name,
s.department,
s.math_score,
d.dept_name,
d.hod
FROM students AS s
LEFT JOIN departments AS d
ON s.dept_code = d.dept_code

UNION ALL

SELECT
s.name,
s.department,
s.math_score,
d.dept_name,
d.hod
FROM departments AS d
LEFT JOIN students AS s
ON s.dept_code = d.dept_code
WHERE s.dept_code IS NULL
"""

run_query(query_full_outer_simulated)
```
