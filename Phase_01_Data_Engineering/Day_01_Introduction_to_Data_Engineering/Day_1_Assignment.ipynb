```python
import pandas as pd

print("Successfully imported library")
df = pd.read_csv("student_performance.csv")
df["total_score"] = (
    df["math_score"] + df["programming_score"]
) / 2

print("\nPRACTICE QUESTIONS")
print("=" * 50)

# 1
ece_students = df[df["department"] == "ECE"].shape[0]
print(f"1. Students from Electronics Department : {ece_students}")

```

```python
# 2
female_attendance = df[df["gender"] == "Female"]["attendance_percentage"].mean()
print(f"2. Average attendance of female students : {female_attendance:.2f}")
```

```python
# 3
lowest_prog_student = df.loc[df["programming_score"].idxmin()]

print("\n3. Student with lowest programming score")
print(f"   Name  : {lowest_prog_student['name']}")
print(f"   Score : {lowest_prog_student['programming_score']}")

```

```python
# 4
above_80_math = df[df["math_score"] > 80].shape[0]
print(f"\n4. Students scored above 80 in math : {above_80_math}")

```

```python
# 5
high_attendance_avg = df[df["attendance_percentage"] > 90]["total_score"].mean()

print(f"\n5. Average total score (attendance > 90%) : {high_attendance_avg:.2f}")
```
