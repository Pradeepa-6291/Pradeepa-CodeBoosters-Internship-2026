```python
import requests
import pandas as pd
```

```python
url = "https://api.open-meteo.com/v1/forecast"

params = {
    "latitude": 13.15,
    "longitude": 80.25,
    "current_weather": True
}

response = requests.get(url, params=params)

weather_json = response.json()
print(weather_json)
```

```python
import json

with open("raw_weather.json","w") as f:
    json.dump(weather_json,f)
```

```python
df = pd.json_normalize(
    weather_json["current_weather"]
)
display(df)
```

```python
df.fillna(0,inplace=True)
```

```python
df["temperature"] = df["temperature"].astype(float)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.to_csv(
    "cleaned_weather.csv",
    index=False
)
```

**PROJECT B**

```python
import pandas as pd

```

```python
df = pd.read_csv("employee.csv")

print("Before Cleaning:", len(df))

df.drop_duplicates(inplace=True)
```

```python
df["basic_salary"] = df["basic_salary"].fillna(
    df["basic_salary"].mean()
)

df["bonus"] = df["bonus"].fillna(0)

df["gross_salary"] = (
    df["basic_salary"] +
    df["bonus"]
)

df["tax"] = df["gross_salary"] * 0.10

df["net_salary"] = (
    df["gross_salary"] -
    df["tax"]
)
```

```python
summary = {
    "Total Employees": len(df),
    "Average Gross Salary": round(df["gross_salary"].mean(), 2),
    "Average Tax": round(df["tax"].mean(), 2),
    "Average Net Salary": round(df["net_salary"].mean(), 2)
}

summary_df = pd.DataFrame(
    list(summary.items()),
    columns=["Metric", "Value"]
)
```

```python
df.to_excel(
    "cleaned_employee_salary.xlsx",
    index=False
)

summary_df.to_excel(
    "salary_summary_report.xlsx",
    index=False
)

print(summary_df)
```
